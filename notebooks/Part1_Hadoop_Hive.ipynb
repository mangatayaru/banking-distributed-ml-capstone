## HDFS ingestion (run in PowerShell after starting the Docker Compose stack)
    ```powershell
    docker cp data/bank.csv namenode:/tmp/bank.csv
    docker exec -it namenode hdfs dfs -mkdir -p /data/banking/raw
    docker exec -it namenode hdfs dfs -put -f /tmp/bank.csv /data/banking/raw/bank.csv
    docker exec -it namenode hdfs dfs -ls /data/banking/raw
    ```

    HDFS stores file blocks across DataNodes. Hive is not the storage engine: it stores table metadata in a metastore and lets analysts query files using HiveQL.
